```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

# -----------------------------
# Step 1: Select relevant features
# -----------------------------

features = [
    'price',
    'freight_value',
    'payment_installments',
    'payment_value',
    'on_time_delivery',
    'approval_hours',
    'status_approved',
    'status_canceled',
    'status_delivered',
    'status_invoiced',
    'status_processing',
    'status_shipped',
    'status_unavailable',
    'pay_boleto',
    'pay_credit_card',
    'pay_debit_card',
    'pay_voucher',
    'extra_charge',
    'demand_High Demand',
    'demand_Low Demand',
    'demand_Medium Demand',
    'demand_Very High Demand',
    'demand_Very Low Demand'
]

X = df_encoded[features]

# Bad = 0 → review scores 1, 2
# Good = 1 → review scores 3, 4, 5
y = (df_encoded['review_score'] >= 3).astype(int)

# -----------------------------
# Step 2: Train/Test Split
# 70% Train / 30% Test
# -----------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42,
    stratify=y
)

# -----------------------------
# Step 3: Feature Scaling
# -----------------------------

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# -----------------------------
# Step 4: Calculate Class Weights
# -----------------------------

classes = np.unique(y_train)

weights = compute_class_weight(
    class_weight='balanced',
    classes=classes,
    y=y_train
)

class_weights = dict(zip(classes, weights))

print("Class Weights:")
print(class_weights)

# -----------------------------
# Step 5: Neural Network
# -----------------------------

model = keras.Sequential([
    layers.Input(shape=(X_train_scaled.shape[1],)),

    layers.Dense(256, activation='relu'),
    layers.Dropout(0.3),

    layers.Dense(128, activation='relu'),
    layers.Dropout(0.3),

    layers.Dense(64, activation='relu'),
    layers.Dropout(0.3),

    layers.Dense(32, activation='relu'),
    layers.Dropout(0.3),

    layers.Dense(16, activation='relu'),
    layers.Dropout(0.3),

    layers.Dense(8, activation='relu'),
    layers.Dropout(0.3),

    layers.Dense(1, activation='sigmoid')
])

# -----------------------------
# Step 6: Compile
# -----------------------------

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()

# -----------------------------
# Step 7: Train
# -----------------------------

history = model.fit(
    X_train_scaled,
    y_train,
    epochs=10,
    batch_size=32,
    validation_split=0.2,
    class_weight=class_weights,
    verbose=1
)

# -----------------------------
# Step 8: Evaluate
# -----------------------------

loss, accuracy = model.evaluate(
    X_test_scaled,
    y_test,
    verbose=0
)

print("\n" + "=" * 65)
print("TensorFlow/Keras Binary Classification - Class Weighted")
print("=" * 65)
print("Target: Bad (1,2) vs Good (3,4,5)")
print("Features: Order, Payment, Delivery, Status & Demand")
print("Train/Test Split: 70/30")
print("Stratified Split: Yes")
print("Feature Scaling: StandardScaler")
print("Architecture: 6 Hidden Dense Layers")
print("Hidden Activation: ReLU")
print("Dropout: 0.3")
print("Output Layer: 1 Neuron + Sigmoid")
print("Optimizer: Adam")
print("Loss Function: Binary Crossentropy")
print("Class Imbalance Handling: Balanced Class Weights")
print("Epochs: 10 | Batch Size: 32")
print(f"Test Accuracy: {accuracy:.4f}")
print("=" * 65)

# -----------------------------
# Step 9: Predictions
# -----------------------------

y_pred_prob = model.predict(
    X_test_scaled,
    verbose=0
)

y_pred = (y_pred_prob >= 0.5).astype(int).ravel()

# -----------------------------
# Step 10: Classification Report
# -----------------------------

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        target_names=['Bad', 'Good']
    )
)
```

```python
import lightgbm as lgb
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score


# ============================================================
# Step 1: Create better delivery / experience features
# ============================================================

df = df_encoded.copy()

# Convert datetime columns if they exist
date_columns = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]

for col in date_columns:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors='coerce')


# ------------------------------------------------------------
# Approval time
# ------------------------------------------------------------

if (
    'order_purchase_timestamp' in df.columns
    and 'order_approved_at' in df.columns
):
    df['approval_hours_new'] = (
        df['order_approved_at']
        - df['order_purchase_timestamp']
    ).dt.total_seconds() / 3600


# ------------------------------------------------------------
# Actual delivery duration
# ------------------------------------------------------------

if (
    'order_purchase_timestamp' in df.columns
    and 'order_delivered_customer_date' in df.columns
):
    df['delivery_days'] = (
        df['order_delivered_customer_date']
        - df['order_purchase_timestamp']
    ).dt.total_seconds() / 86400


# ------------------------------------------------------------
# Delivery delay
#
# Positive = late
# Negative = early
# Zero     = exactly on time
# ------------------------------------------------------------

if (
    'order_delivered_customer_date' in df.columns
    and 'order_estimated_delivery_date' in df.columns
):
    df['delivery_delay_days'] = (
        df['order_delivered_customer_date']
        - df['order_estimated_delivery_date']
    ).dt.total_seconds() / 86400


# ------------------------------------------------------------
# Carrier shipping time
# ------------------------------------------------------------

if (
    'order_delivered_carrier_date' in df.columns
    and 'order_delivered_customer_date' in df.columns
):
    df['carrier_to_customer_days'] = (
        df['order_delivered_customer_date']
        - df['order_delivered_carrier_date']
    ).dt.total_seconds() / 86400


# ============================================================
# Step 2: Select features
# ============================================================

features = [
    # Cost
    'price',
    'freight_value',
    'payment_value',
    'payment_installments',
    'extra_charge',

    # Existing experience feature
    'on_time_delivery',
    'approval_hours',

    # NEW delivery features
    'approval_hours_new',
    'delivery_days',
    'delivery_delay_days',
    'carrier_to_customer_days',

    # Demand
    'demand_High Demand',
    'demand_Low Demand',
    'demand_Medium Demand',
    'demand_Very High Demand',
    'demand_Very Low Demand',

    # Payment
    'pay_boleto',
    'pay_credit_card',
    'pay_debit_card',
    'pay_voucher',

    # Status
    'status_canceled',
    'status_delivered',
    'status_processing',
    'status_shipped',
    'status_invoiced',
    'status_approved',
    'status_unavailable'
]


# Keep only columns that actually exist
features = [
    col for col in features
    if col in df.columns
]

X = df[features].copy()


# ============================================================
# Step 3: Handle missing values
# ============================================================

X = X.replace(
    [float('inf'), float('-inf')],
    pd.NA
)

X = X.fillna(
    X.median(numeric_only=True)
)


# ============================================================
# Step 4: Create target
# ============================================================

# 1,2 -> Bad       = 0
# 3   -> Undecided = 1
# 4,5 -> Good      = 2

y = df['review_score'].map({
    1: 0,
    2: 0,
    3: 1,
    4: 2,
    5: 2
})


# Remove rows where target is missing
valid = y.notna()

X = X.loc[valid]
y = y.loc[valid]


# ============================================================
# Step 5: Train/Test Split
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=10,
    stratify=y
)


# ============================================================
# Step 6: LightGBM
# ============================================================

model = lgb.LGBMClassifier(
    objective='multiclass',
    num_class=3,

    n_estimators=300,
    learning_rate=0.05,

    num_leaves=31,
    max_depth=-1,

    subsample=0.8,
    colsample_bytree=0.8,

    class_weight={
        0: 2.0,   # Bad
        1: 4.0,   # Undecided
        2: 0.5    # Good
    },

    random_state=10,
    n_jobs=-1
)


# ============================================================
# Step 7: Train
# ============================================================

model.fit(
    X_train,
    y_train
)


# ============================================================
# Step 8: Predict
# ============================================================

y_pred = model.predict(X_test)


# ============================================================
# Step 9: Results
# ============================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

print("\n" + "=" * 75)
print("LightGBM Multiclass Classification")
print("Enhanced Customer-Experience Features")
print("=" * 75)

print("Target:")
print("1,2 -> Bad")
print("3   -> Undecided")
print("4,5 -> Good")

print("\nFeatures Used:", len(features))

print("\nNew Experience Features:")

for feature in [
    'approval_hours_new',
    'delivery_days',
    'delivery_delay_days',
    'carrier_to_customer_days'
]:
    if feature in features:
        print(" -", feature)

print("\nTrain/Test Split: 70/30")
print("Stratified Split: Yes")

print("\nModel:")
print("LightGBM Gradient Boosted Decision Trees")
print("Objective: Multiclass")
print("Classes: 3")

print("\nClass Weights:")
print({
    'Bad': 2.0,
    'Undecided': 4.0,
    'Good': 0.5
})

print("\nEstimators: 300")
print("Learning Rate: 0.05")
print("Number of Leaves: 31")

print(f"\nTest Accuracy: {accuracy:.4f}")

print("=" * 75)


# ============================================================
# Step 10: Classification Report
# ============================================================

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred,
        target_names=[
            'Bad',
            'Undecided',
            'Good'
        ]
    )
)


# ============================================================
# Step 11: Prediction Distribution
# ============================================================

print("\nActual Distribution:")

print(
    pd.Series(y_test)
    .map({
        0: 'Bad',
        1: 'Undecided',
        2: 'Good'
    })
    .value_counts()
)


print("\nPredicted Distribution:")

print(
    pd.Series(y_pred)
    .map({
        0: 'Bad',
        1: 'Undecided',
        2: 'Good'
    })
    .value_counts()
)


# ============================================================
# Step 12: Actual Rating vs Prediction
# ============================================================

result = pd.DataFrame({
    'Actual Review Score':
        df.loc[
            X_test.index,
            'review_score'
        ],

    'Predicted Class':
        y_pred
})

result['Predicted Class'] = result[
    'Predicted Class'
].map({
    0: 'Bad',
    1: 'Undecided',
    2: 'Good'
})


print("\nActual Review Score vs Predicted Class:")

print(
    pd.crosstab(
        result['Actual Review Score'],
        result['Predicted Class']
    )
)


# ============================================================
# Step 13: Feature Importance
# ============================================================

importance = pd.DataFrame({
    'Feature': features,
    'Importance': model.feature_importances_
})

importance = importance.sort_values(
    'Importance',
    ascending=False
)

print("\nLightGBM Feature Importance:")

print(importance.to_string(index=False))
```
