# combinbe tables

```python
customers.shape
```

```python
print("=== TESTING ROW COUNTS AT EACH STEP ===\n")

# Step 1: customers + orders
step1 = customers.merge(orders, on='customer_id', how='inner')
print(f"Step 1 - customers + orders: {step1.shape[0]} rows")
print(f"  All {len(customers)} customers have orders ✓")

# Step 2: + order_items
step2 = step1.merge(order_items, on='order_id', how='inner')
print(f"\nStep 2 - + order_items: {step2.shape[0]} rows")
print(f"  Before: {step1.shape[0]} orders")
print(f"  After: {step2.shape[0]} rows (INCREASED because one order can have multiple items)")
print(f"  Total items: {len(order_items)}")
print(f"  Orders with items: {order_items['order_id'].nunique()}")
print(f"  Orders WITHOUT items: {step1.shape[0] - order_items['order_id'].nunique()}")

# Step 3: + products
step3 = step2.merge(products, on='product_id', how='inner')
print(f"\nStep 3 - + products: {step3.shape[0]} rows")
print(f"  Before: {step2.shape[0]} order items")
print(f"  After: {step3.shape[0]} rows")
print(f"  Items WITH valid products: {step2[step2['product_id'].isin(products['product_id'])].shape[0]}")
print(f"  Items WITHOUT valid products: {step2.shape[0] - step3.shape[0]}")

# Step 4: + sellers
step4 = step3.merge(sellers, on='seller_id', how='inner')
print(f"\nStep 4 - + sellers: {step4.shape[0]} rows")
print(f"  Before: {step3.shape[0]} rows")
print(f"  After: {step4.shape[0]} rows")
print(f"  Items WITH valid sellers: {step3[step3['seller_id'].isin(sellers['seller_id'])].shape[0]}")
print(f"  Items WITHOUT valid sellers: {step3.shape[0] - step4.shape[0]}")

# Step 5: + order_payments
step5 = step4.merge(order_payments, on='order_id', how='inner')
print(f"\nStep 5 - + order_payments: {step5.shape[0]} rows")
print(f"  Before: {step4.shape[0]} rows")
print(f"  After: {step5.shape[0]} rows (INCREASED because one order can have multiple payments)")
print(f"  Orders WITH payments: {order_payments['order_id'].nunique()}")
print(f"  Orders WITHOUT payments: {step1.shape[0] - order_payments['order_id'].nunique()}")

# Step 6: + order_reviews
step6 = step5.merge(order_reviews, on='order_id', how='inner')
print(f"\nStep 6 - + order_reviews: {step6.shape[0]} rows")
print(f"  Before: {step5.shape[0]} rows")
print(f"  After: {step6.shape[0]} rows")
print(f"  Orders WITH reviews: {order_reviews['order_id'].nunique()}")
print(f"  Orders WITHOUT reviews: {step1.shape[0] - order_reviews['order_id'].nunique()}")

print("\n" + "="*60)
print("FINAL SUMMARY:")
print("="*60)
print(f"Starting orders: {step1.shape[0]}")
print(f"Final rows: {step6.shape[0]}")
print(f"Rows INCREASED due to multiple items/payments per order")
print(f"Rows DROPPED due to missing products/sellers/reviews")
print("="*60)
```

```python
final_df = (customers
    .merge(orders, on='customer_id', how='inner')
    .merge(order_items, on='order_id', how='inner')
    .merge(products, on='product_id', how='inner')
    .merge(sellers, on='seller_id', how='inner')
    .merge(order_payments, on='order_id', how='inner')
    .merge(order_reviews, on='order_id', how='inner')
)

print(f"Final dataset: {final_df.shape}")
print(f"Columns: {final_df.columns.tolist()}")
```
