## 🧹 Dataset Preprocessing — Summary

* Loaded **8,693 training** and **4,277 test** passengers.
* Removed `PassengerId` as it is only an identifier.
* Investigated missing values; most rows had **0–1 missing values** and each feature had only ~2% missing data.
* Tested **model-based imputation** for categorical values, but accuracy was only ~72%, so this approach was not used.
* Filled missing categorical values (`HomePlanet`, `CryoSleep`, `Destination`, `VIP`) with `"Unknown"`.
* Dropped `Cabin` and `Name`: `Cabin` provided limited individual information, while `Name` was mostly unique.
* Filled missing numeric values (`Age`, `RoomService`, `FoodCourt`, `ShoppingMall`, `Spa`, `VRDeck`) with the **median**.
* Converted categorical features to **numeric dummy variables** using one-hot encoding.
* Verified that train/test feature counts matched correctly and that **no missing values remained**.
* Saved the processed datasets as `.csv` and `.pkl` files in `cdata/`.

### 🔎 Key Experiment

Model-based filling of categorical missing values was tested using a Random Forest, but the validation accuracy was only **~72.3%**. Therefore, `"Unknown"` was used instead, preserving the fact that the original value was missing.
